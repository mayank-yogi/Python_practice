# 1. Smallest subarray with sum greater than target

```python
l1 = [2,3,1,2,4,3]
target = 7

small = len(l1) + 1

for i in range(len(l1)):

    total = 0

    for j in range(i,len(l1)):

        total = total + l1[j]

        if total > target:

            length = j - i + 1

            if length < small:
                small = length

            break

if small == len(l1) + 1:
    print(0)
else:
    print(small)
```

# 2. Sort a list using Merge Sort

```python
l1 = [5,2,8,1,3]
lists = []

for i in l1:
    lists.append([i])

while len(lists) > 1:

    new = []

    for i in range(0,len(lists),2):

        if i + 1 < len(lists):

            l1 = lists[i]
            l2 = lists[i+1]

            temp = []
            a = 0
            b = 0

            while a < len(l1) and b < len(l2):

                if l1[a] < l2[b]:
                    temp.append(l1[a])
                    a += 1
                else:
                    temp.append(l2[b])
                    b += 1

            while a < len(l1):
                temp.append(l1[a])
                a += 1

            while b < len(l2):
                temp.append(l2[b])
                b += 1

            new.append(temp)

        else:
            new.append(lists[i])

    lists = new

print(lists[0])
```

# 3. Maximum product subarray

```python
l1 = [2,3,-2,4]

max_product = l1[0]

for i in range(len(l1)):

    product = 1

    for j in range(i,len(l1)):

        product = product * l1[j]

        if product > max_product:
            max_product = product

print(max_product)
```

# 4. First missing positive integer 

```python
l1 = [3,4,-1,1]

n = len(l1)
missing = 1

for i in range(1,n+2):

    found = False

    for j in l1:

        if j == i:
            found = True
            break

    if found == False:
        missing = i
        break

print(missing)
```

# 5. Rearrange positive and negative numbers alternately

```python
l1 = [1,2,-3,-4,5,-6]

positive = []
negative = []

for i in l1:

    if i >= 0:
        positive.append(i)
    else:
        negative.append(i)

res = []

p = 0
n = 0

for i in range(len(l1)):

    if i % 2 == 0:

        if p < len(positive):
            res.append(positive[p])
            p += 1

        elif n < len(negative):
            res.append(negative[n])
            n += 1

    else:

        if n < len(negative):
            res.append(negative[n])
            n += 1

        elif p < len(positive):
            res.append(positive[p])
            p += 1

print(res)
```

# t6. Majority Element

```python
l1 = [2,2,1,1,1,2,2]

majority = -1

for i in l1:

    count = 0

    for j in l1:

        if i == j:
            count += 1

    if count > len(l1) / 2:
        majority = i
        break

print(majority)
```

# 7. Find the next permutation

```python
l1 = [1,2,3]

i = len(l1) - 2

for x in range(len(l1)-1):
    if l1[i] < l1[i+1]:
        break

    i -= 1

if i >= 0:

    j = len(l1) - 1

    for x in range(len(l1)):
        if l1[j] > l1[i]:
            break
        j -= 1

    temp = l1[i]
    l1[i] = l1[j]
    l1[j] = temp

left = i + 1
right = len(l1) - 1

for x in range((right-left+1)//2):
    temp = l1[left]
    l1[left] = l1[right]
    l1[right] = temp

    left += 1
    right -= 1

print(l1)
```

# 8. Shortest subarray with the same degree as original list

```python
l1 = [1,2,2,3,1]

freq = {}
for i in l1:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

degree = 0

for i in freq:
    if freq[i] > degree:
        degree = freq[i]

small = len(l1) + 1

for i in freq:
    if freq[i] == degree:
        first = -1
        last = -1

        for j in range(len(l1)):
            if l1[j] == i:

                if first == -1:
                    first = j
                last = j

        length = last - first + 1

        if length < small:
            small = length

print(small)
```

# 9. Sliding window maximum

```python
l1 = [1,3,-1,-3,5,3,6,7]
k = 3

res = []
for i in range(len(l1)-k+1):
    max = l1[i]

    for j in range(i,i+k):
        if l1[j] > max:
            max = l1[j]

    res.append(max)

print(res)
```

# 10. Generate all possible subsets — Power Set

```python
l1 = [1,2,3]
res = [[]]
for i in l1:
    new = []

    for j in res:
        temp = j + [i]
        new.append(temp)

    for j in new:
        res.append(j)

print(res)
```
