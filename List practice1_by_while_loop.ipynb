# Q1 reverse a list 

```python
# reverse a list

l1 = [1,2,3,4]
l2 = []

i = 0

while i < 4:
    l2 = [l1[i]] + l2
    i += 1

print(l2)
```

# 2. Find max element 

```python
l1 = [1,2,3,4]

max = l1[0]
i = 0

while i < 4:
    if l1[i] > max:
        max = l1[i]
    i += 1

print(max)
```

# 4. Remove duplicates 

```python
# remove duplicates

l3 = [1,2,3,44,2,4,1,34]
l4 = []

i = 0

while i < 8:
    if l3[i] not in l4:
        l4 = l4 + [l3[i]]
    i += 1

print(l4)
```

# 5. Second largest element 

```python
l1 = [1,2,3,444,566,223]

max = l1[0]
i = 0

while i < 6:
    if l1[i] > max:
        max = l1[i]
    i += 1

sec = l1[0]
i = 0

while i < 6:
    if l1[i] > sec and l1[i] < max:
        sec = l1[i]
    i += 1

print(sec)
```

# 6. merge list 

```python
l1 = [1, 3, 5, 7]
l2 = [2, 4, 6, 8]

res = []
i = 0
j = 0

while i < len(l1) and j < len(l2):
    if l1[i] < l2[j]:
        res.append(l1[i])
        i += 1
    else:
        res.append(l2[j])
        j += 1

# Add remaining elements
while i < len(l1):
    res.append(l1[i])
    i += 1

while j < len(l2):
    res.append(l2[j])
    j += 1

print(res)
```

```python
# rotate a list k times

l1 = [1,2,3,4,4,5,6,7,8,8]

num = int(input("enter how many times u want to rotate the list :"))

for i in l1:
    if num >= 10:
        num = num - 10

l2 = []

for i in l1:
    l2 = l2 + [i]

print(l2[num:] + l2[:num])
```

# 8. Palindrome 

```python
my_list = [1, 2, 3, 2, 1]

is_palindrome = True

left = 0
right = len(my_list) - 1

while left < right:
    if my_list[left] != my_list[right]:
        is_palindrome = False
        break

    left += 1
    right -= 1

if is_palindrome:
    print("Palindrome")
else:
    print("Not Palindrome")

```

# 9. count freq 

```python
# count freq

data = [1,2,3,4,4,55,6,6,7]
freq_table = {}

i = 0

while i < 9:
    x = data[i]

    if x in freq_table:
        freq_table[x] += 1
    else:
        freq_table[x] = 1

    i += 1

print(freq_table)
```

# 10 missing no in range 1 to n 

```python
# missing number in range 1 to n

my_list = [1,2,3,5,6]

n = 6

total = 0
actual = 0

i = 1

while i <= n:
    total = total + i
    i += 1

i = 0

while i < 5:
    actual = actual + my_list[i]
    i += 1

missing = total - actual

print("Missing number:", missing)
```
