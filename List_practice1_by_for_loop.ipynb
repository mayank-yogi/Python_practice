```python
# palindrome

my_list = [1,2,3,2,1]

rev = []

for i in my_list:
    rev = [i] + rev

if my_list == rev:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# count freq
data = [1,2,3,4,4,55,6,6,7]
freq_table = {}
for x in data:
    if x in freq_table:
        freq_table[x] += 1
    else:
        freq_table[x] = 1
print(freq_table)
```

```python
# missing number in range 1 to n 
my_list = [1, 2, 3, 5, 6]

n = 6

total = 0
actual = 0

# Calculate 1 + 2 + ... + n
for i in range(1, n + 1):
    total = total + i

# Calculate sum of elements in list
for i in range(len(my_list)):
    actual = actual + my_list[i]

missing = total - actual

print("Missing number:", missing)
```

```python
# sort a list 
my_list = [5, 2, 8, 1, 3]

n = len(my_list)

for i in range(n):
    for j in range(0, n - i - 1):

        if my_list[j] > my_list[j + 1]:
            temp = my_list[j]
            my_list[j] = my_list[j + 1]
            my_list[j + 1] = temp

print(my_list)
```
