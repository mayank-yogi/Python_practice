# 1. Smallest subarray with sum greater than target

```python
l1 = [2,3,1,2,4,3]
target = 7
small = len(l1) + 1
i = 0

while i < len(l1):
    total = 0
    j = i

    while j < len(l1):
        total = total + l1[j]

        if total > target:
            length = j - i + 1

            if length < small:
                small = length

            break

        j += 1

    i += 1

if small == len(l1) + 1:
    print(0)
else:
    print(small)
```

# 2. Merge Sort manually

```python
l1 = [5,2,8,1,3]
lists = []
i = 0

while i < len(l1):
    lists.append([l1[i]])
    i += 1

while len(lists) > 1:
    new = []
    i = 0

    while i < len(lists):
        if i + 1 < len(lists):
            l2 = lists[i]
            l3 = lists[i+1]
            temp = []
            a = 0
            b = 0

            while a < len(l2) and b < len(l3):
                if l2[a] < l3[b]:
                    temp.append(l2[a])
                    a += 1
                else:
                    temp.append(l3[b])
                    b += 1

            while a < len(l2):
                temp.append(l2[a])
                a += 1

            while b < len(l3):
                temp.append(l3[b])
                b += 1
            new.append(temp)

        else:
            new.append(lists[i])

        i += 2
    lists = new

print(lists[0])
```

# 3. Maximum product subarray

```python
l1 = [2,3,-2,4]
max_product = l1[0]

i = 0

while i < len(l1):
    product = 1
    j = i

    while j < len(l1):
        product = product * l1[j]
        if product > max_product:
            max_product = product
        j += 1
    i += 1

print(max_product)
```

# 4. First missing positive integer

```python
l1 = [3,4,-1,1]
n = len(l1)
missing = 1
i = 1

while i <= n + 1:
    found = False
    j = 0
    while j < len(l1):
        if l1[j] == i:
            found = True
            break
        j += 1

    if found == False:
        missing = i
        break
    i += 1

print(missing)
```

# 5. Rearrange positive and negative numbers alternately

```python
l1 = [1,2,-3,-4,5,-6]
positive = []
negative = []
i = 0

while i < len(l1):
    if l1[i] >= 0:
        positive.append(l1[i])
    else:
        negative.append(l1[i])
    i += 1
    
res = []
p = 0
n = 0
i = 0

while i < len(l1):
    if i % 2 == 0:
        if p < len(positive):
            res.append(positive[p])
            p += 1
        elif n < len(negative):
            res.append(negative[n])
            n += 1
    else:
        if n < len(negative):
            res.append(negative[n])
            n += 1

        elif p < len(positive):
            res.append(positive[p])
            p += 1
    i += 1
print(res)
```

# 6. Majority element

```python
l1 = [2,2,1,1,1,2,2]
majority = -1
i = 0

while i < len(l1):
    count = 0
    j = 0

    while j < len(l1):
        if l1[i] == l1[j]:
            count += 1
        j += 1

    if count > len(l1) / 2:
        majority = l1[i]
        break
    i += 1

print(majority)
```

# 7. Next permutation

```python
l1 = [1,2,3]
i = len(l1) - 2

while i >= 0:
    if l1[i] < l1[i+1]:
        break
    i -= 1

if i >= 0:
    j = len(l1) - 1

    while j > i:
        if l1[j] > l1[i]:
            break
        j -= 1
    temp = l1[i]
    l1[i] = l1[j]
    l1[j] = temp

left = i + 1
right = len(l1) - 1

while left < right:
    temp = l1[left]
    l1[left] = l1[right]
    l1[right] = temp
    left += 1
    right -= 1

print(l1)
```

# 8. Shortest subarray with same degree

```python
l1 = [1,2,2,3,1]
freq = {}
i = 0

while i < len(l1):
    if l1[i] in freq:
        freq[l1[i]] += 1
    else:
        freq[l1[i]] = 1
    i += 1

degree = 0
keys = []

for i in freq:
    keys.append(i)
i = 0

while i < len(keys):
    if freq[keys[i]] > degree:
        degree = freq[keys[i]]
    i += 1

small = len(l1) + 1
i = 0

while i < len(keys):
    if freq[keys[i]] == degree:
        first = -1
        last = -1
        j = 0

        while j < len(l1):
            if l1[j] == keys[i]:
                if first == -1:
                    first = j
                last = j
            j += 1
        length = last - first + 1

        if length < small:
            small = length
    i += 1

print(small)
```

## 9. Sliding window maximum 

```python
l1 = [1,3,-1,-3,5,3,6,7]
k = 3
res = []
i = 0

while i <= len(l1) - k:
    max = l1[i]
    j = i
    while j < i + k:
        if l1[j] > max:
            max = l1[j]
        j += 1
    res.append(max)
    i += 1

print(res)
```

# 10. Power Set — all possible subsets

```python
l1 = [1,2,3]
res = [[]]
i = 0

while i < len(l1):
    new = []
    j = 0
    while j < len(res):
        temp = res[j] + [l1[i]]
        new.append(temp)
        j += 1
    j = 0

    while j < len(new):
        res.append(new[j])
        j += 1
    i += 1

print(res)
```
