# 1. Find all pairs with sum target 

```python
l1 = [2,4,3,5,7,8]
target = 7

pairs = []

i = 0

while i < 6:
    j = i + 1

    while j < 6:
        if l1[i] + l1[j] == target:
            pairs.append([l1[i],l1[j]])
        j += 1

    i += 1

print(pairs)
```

# 2. Maximum subarray sum

```python
l1 = [-2,1,-3,4,-1,2,1,-5,4]

current = l1[0]
max = l1[0]

i = 1

while i < 9:

    if current + l1[i] > l1[i]:
        current = current + l1[i]
    else:
        current = l1[i]

    if current > max:
        max = current

    i += 1

print(max)
```

# 3. Intersection of two lists

```python
l1 = [1,2,3,4,5]
l2 = [1,2,5,6,7]

res = []

i = 0

while i < 5:

    j = 0

    while j < 5:

        if l1[i] == l2[j]:
            if l1[i] not in res:
                res.append(l1[i])

        j += 1

    i += 1

print(res)
```

# 4. Move all zeros to the end

```python
l1 = [0,1,0,33,2,0]
res = []

zero = 0
i = 0

while i < 5:

    if l1[i] != 0:
        res.append(l1[i])
    else:
        zero += 1

    i += 1

i = 0

while i < zero:
    res.append(0)
    i += 1

print(res)
```

# 5. Flatten a nested list without recursion

```python
l1 = [11,[22,33],[44,[55,76]]]
res = []

i = 0

while i < 3:

    if type(l1[i]) == list:

        j = 0

        while j < len(l1[i]):

            if type(l1[i][j]) == list:

                k = 0

                while k < len(l1[i][j]):
                    res.append(l1[i][j][k])
                    k += 1

            else:
                res.append(l1[i][j])

            j += 1

    else:
        res.append(l1[i])

    i += 1

print(res)
```

# 6. Longest Increasing Subsequence

```python
# longest increasing subsequence

l1 = [10,9,2,5,3,7,101,18]

n = 8

dp = [1] * n

i = 0

while i < n:

    j = 0

    while j < i:

        if l1[j] < l1[i]:

            if dp[j] + 1 > dp[i]:
                dp[i] = dp[j] + 1

        j += 1

    i += 1

max = dp[0]
pos = 0
i = 0

while i < n:

    if dp[i] > max:
        max = dp[i]
        pos = i

    i += 1

res = []
current = l1[pos]

res.append(current)

i = pos - 1

while i >= 0:

    if l1[i] < current and dp[i] == max-1:
        res = [l1[i]] + res
        current = l1[i]
        max = max - 1

    i -= 1

print(res)
```

# 7. Group anagrams 

```python
words = ["eat","tea","tan","ate","nat","bat"]

groups = []

i = 0

while i < 6:

    word = words[i]
    found = False

    j = 0

    while j < len(groups):

        first = groups[j][0]

        count1 = {}
        count2 = {}

        k = 0

        while k < len(first):

            ch = first[k]

            if ch in count1:
                count1[ch] += 1
            else:
                count1[ch] = 1

            k += 1

        k = 0

        while k < len(word):

            ch = word[k]

            if ch in count2:
                count2[ch] += 1
            else:
                count2[ch] = 1

            k += 1

        if count1 == count2:
            groups[j].append(word)
            found = True
            break

        j += 1

    if found == False:
        groups.append([word])

    i += 1

print(groups)
```

# 8. Generate all permutations of a list

```python
l1 = [1,2,3]

res = [[]]

i = 0

while i < len(l1):

    item = l1[i]
    new = []

    j = 0

    while j < len(res):

        p = res[j]
        k = 0

        while k <= len(p):

            temp = p[:k] + [item] + p[k:]
            new.append(temp)

            k += 1

        j += 1

    res = new
    i += 1

print(res)
```

# 9. Equilibrium index

```python
l1 = [1,3,5,2,2]

total = 0
i = 0

while i < 5:
    total = total + l1[i]
    i += 1

left = 0
index = -1
i = 0

while i < 5:

    right = total - left - l1[i]

    if left == right:
        index = i
        break

    left = left + l1[i]
    i += 1

print(index)
```

# 10. Split a list into chunks of size k

```python

l1 = [1,2,3,4,5,6,7]
k = 3

res = []
temp = []

i = 0

while i < 7:

    temp.append(l1[i])

    if len(temp) == k:
        res.append(temp)
        temp = []

    i += 1

if len(temp) != 0:
    res.append(temp)

print(res)
```
