# 1. Find all pairs in a list that sum up to a target value.

```python
l1 = [2,4,3,5,7,8]
target = 9

pairs = []

for i in range(6):
    for j in range(i+1,6):
        if l1[i] + l1[j] == target:
            pairs.append([l1[i],l1[j]])

print(pairs)
```

# 2. Max sub array sum kdane's algo 

```python
l1 = [-2,1,-3,4,-1,2,1,-5,4]

current = l1[0]
max = l1[0]

for i in range(1,9):

    if current + l1[i] > l1[i]:
        current = current + l1[i]
    else:
        current = l1[i]

    if current > max:
        max = current

print(max)
```

# 3. Intersection of 2 lists 

```python

l1 = [1,2,3,4,5]
l2 = [3,4,5,6,7]

res = []

for i in l1:
    for j in l2:
        if i == j:
            if i not in res:
                res.append(i)

print(res)
```

# 4. Move all zeroes to end 

```python
l1 = [0,1,0,3,12]
res = []

zero = 0

for i in l1:

    if i != 0:
        res.append(i)
    else:
        zero += 1

for i in range(zero):
    res.append(0)

print(res)
```

# 5. Flatten nested list without using  recursion 

```python
l1 = [1,[2,3],[4,[5,6]]]
res = []

for i in l1:

    if type(i) == list:

        for j in i:

            if type(j) == list:
                for k in j:
                    res.append(k)
            else:
                res.append(j)

    else:
        res.append(i)

print(res)
```

# 6. Longest subsequence 

```python
l1 = [10,9,2,5,3,7,101,18]

n = 8

dp = [1] * n
res = []

for i in range(n):

    for j in range(i):

        if l1[j] < l1[i]:

            if dp[j] + 1 > dp[i]:
                dp[i] = dp[j] + 1

max = dp[0]
pos = 0

for i in range(n):

    if dp[i] > max:
        max = dp[i]
        pos = i

current = l1[pos]
res.append(current)

for i in range(pos-1,-1,-1):

    if l1[i] < current and dp[i] == max-1:
        res = [l1[i]] + res
        current = l1[i]
        max = max - 1

print(res)
```

# 7. Group anagrams 

```python
words = ["eat","tea","tan","ate","nat","bat"]

groups = []

for word in words:

    found = False

    for group in groups:

        first = group[0]

        count1 = {}
        count2 = {}

        for ch in first:
            if ch in count1:
                count1[ch] += 1
            else:
                count1[ch] = 1

        for ch in word:
            if ch in count2:
                count2[ch] += 1
            else:
                count2[ch] = 1

        if count1 == count2:
            group.append(word)
            found = True
            break

    if found == False:
        groups.append([word])

print(groups)
```

# 8. Generate all permutations of a list 

```python
l1 = [1,2,3]

res = [[]]

for item in l1:

    new = []

    for p in res:

        for i in range(len(p)+1):

            temp = p[:i] + [item] + p[i:]
            new.append(temp)

    res = new

print(res)
```

# 9. Equilibrium index 

```python

l1 = [1,3,5,2,2]

total = 0

for i in l1:
    total = total + i

left = 0
index = -1

for i in range(5):

    right = total - left - l1[i]

    if left == right:
        index = i
        break

    left = left + l1[i]

print(index)
```

# 10. Split a list into chunks of k size 

```python
l1 = [1,2,3,4,5,6,7]
k = 3

res = []
temp = []

count = 0

for i in l1:

    temp.append(i)
    count += 1

    if count == k:
        res.append(temp)
        temp = []
        count = 0

if count != 0:
    res.append(temp)

print(res)
```
